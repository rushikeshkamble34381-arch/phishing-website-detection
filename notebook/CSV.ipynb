```python
# Define the models
models = {
    "Decision Tree": DecisionTreeClassifier(
        max_depth=10,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=2000
    )
}

# Train each model and collect results
results = []

for name, clf in models.items():
    clf.fit(X_train, y_train)

    predictions = clf.predict(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, pos_label=1, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, pos_label=1, zero_division=0
        ),
        "F1-score": f1_score(
            y_test, predictions, pos_label=1, zero_division=0
        )
    })

results_df = pd.DataFrame(results)

print("Model comparison completed!")
display(results_df.round(4))
```

```python
from urllib.parse import urlparse
import ipaddress
import re

SHORTENERS = {
    "bit.ly", "tinyurl.com", "t.co", "is.gd",
    "shorturl.at", "ow.ly", "buff.ly", "goo.gl"
}

def extract_basic_features(url):
    url = url.strip()

    # Add a scheme if the user omitted it
    if not re.match(r"^[a-zA-Z][a-zA-Z0-9+.-]*://", url):
        url = "https://" + url

    parsed = urlparse(url)
    domain = (parsed.hostname or "").lower()

    if not domain:
        raise ValueError("Please enter a valid website URL.")

    # Check whether the domain is an IP address
    try:
        ipaddress.ip_address(domain)
        has_ip = -1
    except ValueError:
        has_ip = 1

    # URL length encoding used by the dataset
    length = len(url)
    if length < 54:
        url_length = 1
    elif length <= 75:
        url_length = 0
    else:
        url_length = -1

    # URL shortening service
    shortener = -1 if domain in SHORTENERS else 1

    # @ symbol in URL
    at_symbol = -1 if "@" in url else 1

    # Double slash after the scheme
    slash_position = url.find("//", url.find("://") + 3)
    double_slash = -1 if slash_position != -1 else 1

    # Hyphen in domain
    prefix_suffix = -1 if "-" in domain else 1

    # Number of subdomains (approximation)
    dots = domain.count(".")
    if dots <= 1:
        subdomain = 1
    elif dots == 2:
        subdomain = 0
    else:
        subdomain = -1

    # "https" token in the domain
    https_token = -1 if "https" in domain else 1

    return {
        "having_ip_address": has_ip,
        "url_length": url_length,
        "shortining_service": shortener,
        "having_at_symbol": at_symbol,
        "double_slash_redirecting": double_slash,
        "prefix_suffix": prefix_suffix,
        "having_sub_domain": subdomain,
        "https_token": https_token
    }

print("URL feature extractor created successfully!")
```

```python
import re
import ipaddress
from urllib.parse import urlparse

SHORTENING_SERVICES = {
    "bit.ly", "tinyurl.com", "t.co", "is.gd",
    "ow.ly", "buff.ly", "rebrand.ly", "cutt.ly"
}

def extract_basic_features(url):
    url = url.strip()

    if not url:
        raise ValueError("Please enter a valid URL.")

    if not re.match(r"^[a-zA-Z][a-zA-Z0-9+.-]*://", url):
        url = "https://" + url

    parsed = urlparse(url)
    domain = (parsed.hostname or "").lower()

    if not domain:
        raise ValueError("Could not identify the domain.")

    # Check whether the hostname is an IP address
    try:
        ipaddress.ip_address(domain)
        has_ip = 1
    except ValueError:
        has_ip = -1

    # Basic URL features
    features = {
        "having_ip_address": has_ip,
        "url_length": 1 if len(url) < 54 else (0 if len(url) <= 75 else -1),
        "shortining_service": 1 if domain in SHORTENING_SERVICES else -1,
        "having_at_symbol": 1 if "@" in url else -1,
        "double_slash_redirecting": 1 if "//" in url[8:] else -1,
        "prefix_suffix": 1 if "-" in domain else -1,
        "having_sub_domain": (
            1 if len(domain.split(".")) > 3
            else (0 if len(domain.split(".")) == 3 else -1)
        ),
        "https_token": 1 if "https" in domain else -1
    }

    return features
```

```python
import os
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load the dataset
df = pd.read_csv("../data/phishing.csv")

print("Dataset shape:", df.shape)
print("Target columns:", [c for c in df.columns if c.lower() == "result"])

# Select features supported by our current basic URL extractor
candidate_features = [
    "having_ip_address",
    "url_length",
    "shortining_service",
    "having_at_symbol",
    "double_slash_redirecting",
    "prefix_suffix",
    "having_sub_domain",
    "https_token"
]

features = [c for c in candidate_features if c in df.columns]

print("\nMatching features:", features)

if "result" not in df.columns:
    raise ValueError(
        "Target column 'result' not found. Check the printed column names."
    )

if len(features) < 5:
    raise ValueError(
        "Not enough matching features. Check the dataset columns."
    )

data = df[features + ["result"]].dropna().drop_duplicates().copy()

# Original UCI labels: -1 = phishing, 1 = legitimate
data["result"] = data["result"].map({-1: 1, 1: 0})
data = data.dropna(subset=["result"])
data["result"] = data["result"].astype(int)

X = data[features]
y = data["result"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

url_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

url_model.fit(X_train, y_train)
predictions = url_model.predict(X_test)

print("\nAccuracy:", round(accuracy_score(y_test, predictions), 4))
print("\nClassification report:")
print(classification_report(
    y_test,
    predictions,
    target_names=["Legitimate", "Phishing"],
    zero_division=0
))

# Save the new model separately
os.makedirs("../model", exist_ok=True)

joblib.dump(
    {
        "model": url_model,
        "feature_names": features,
        "target_mapping": {
            "0": "Legitimate",
            "1": "Phishing"
        }
    },
    "../model/url_only_phishing_model.pkl"
)

print("\nSUCCESS: URL-based model saved!")
print("Location:", os.path.abspath(
    "../model/url_only_phishing_model.pkl"
))
```

```python
import re
import ipaddress
import pandas as pd
from urllib.parse import urlparse

SHORTENING_SERVICES = {
    "bit.ly", "tinyurl.com", "t.co", "is.gd",
    "ow.ly", "buff.ly", "rebrand.ly", "cutt.ly"
}

def extract_basic_features(url):
    url = url.strip()

    if not url:
        raise ValueError("Enter a URL.")

    if not re.match(r"^[a-zA-Z][a-zA-Z0-9+.-]*://", url):
        url = "https://" + url

    parsed = urlparse(url)
    domain = (parsed.hostname or "").lower()

    if not domain:
        raise ValueError("Invalid URL: domain not found.")

    try:
        ipaddress.ip_address(domain)
        has_ip = 1
    except ValueError:
        has_ip = -1

    domain_parts = domain.split(".")

    return {
        "having_ip_address": has_ip,
        "url_length": 1 if len(url) < 54 else (0 if len(url) <= 75 else -1),
        "shortining_service": 1 if domain in SHORTENING_SERVICES else -1,
        "having_at_symbol": 1 if "@" in url else -1,
        "double_slash_redirecting": 1 if "//" in url[8:] else -1,
        "prefix_suffix": 1 if "-" in domain else -1,
        "having_sub_domain": (
            1 if len(domain_parts) > 3
            else (0 if len(domain_parts) == 3 else -1)
        ),
        "https_token": 1 if "https" in domain else -1
    }


def predict_url(url):
    features = extract_basic_features(url)

    missing = [
        name for name in feature_names
        if name not in features
    ]

    if missing:
        raise ValueError(
            "Cannot predict reliably. Missing features: "
            + ", ".join(missing)
        )

    input_df = pd.DataFrame(
        [[features[name] for name in feature_names]],
        columns=feature_names
    )

    prediction = int(model.predict(input_df)[0])

    if prediction == 1:
        result = "Phishing"
    else:
        result = "Legitimate"

    return {
        "URL": url,
        "Prediction": result,
        "Warning": "Prototype result only; not a guarantee of website safety."
    }
```

```python
test_urls = [
    "https://google.com",
    "http://example.com",
    "https://login-example.com/verify"
]

for url in test_urls:
    print("\nTesting:", url)

    try:
        result = predict_url(url)
        print("Prediction:", result["Prediction"])
        print(result["Warning"])
    except ValueError as error:
        print("Prediction unavailable:", error)
```

```python

import streamlit as st
import pandas as pd
import joblib
import re
import ipaddress
from urllib.parse import urlparse
from pathlib import Path

st.set_page_config(
    page_title="PhishGuard",
    page_icon="🛡️",
    layout="centered"
)

MODEL_PATH = (
    Path(__file__).resolve().parent
    / "model"
    / "url_only_phishing_model.pkl"
)

SHORTENING_SERVICES = {
    "bit.ly", "tinyurl.com", "t.co", "is.gd",
    "ow.ly", "buff.ly", "rebrand.ly", "cutt.ly"
}

@st.cache_resource
def load_model():
    return joblib.load(MODEL_PATH)

def extract_basic_features(url):
    url = url.strip()

    if not url:
        raise ValueError("Please enter a URL.")

    if not re.match(r"^[a-zA-Z][a-zA-Z0-9+.-]*://", url):
        url = "https://" + url

    parsed = urlparse(url)
    domain = (parsed.hostname or "").lower()

    if not domain:
        raise ValueError("Please enter a valid URL.")

    try:
        ipaddress.ip_address(domain)
        has_ip = 1
    except ValueError:
        has_ip = -1

    parts = domain.split(".")

    return {
        "having_ip_address": has_ip,
        "url_length": (
            1 if len(url) < 54
            else (0 if len(url) <= 75 else -1)
        ),
        "shortining_service": (
            1 if domain in SHORTENING_SERVICES else -1
        ),
        "having_at_symbol": 1 if "@" in url else -1,
        "double_slash_redirecting": (
            1 if "//" in url[8:] else -1
        ),
        "prefix_suffix": 1 if "-" in domain else -1,
        "having_sub_domain": (
            1 if len(parts) > 3
            else (0 if len(parts) == 3 else -1)
        ),
        "https_token": 1 if "https" in domain else -1
    }

st.title("🛡️ PhishGuard")
st.subheader("Phishing Website Detection")
st.write(
    "Enter a website URL to check its basic URL characteristics "
    "with your machine learning model."
)

url = st.text_input(
    "Website URL",
    placeholder="https://example.com"
)

if st.button("Check Website", type="primary", use_container_width=True):
    try:
        bundle = load_model()
        model = bundle["model"]
        feature_names = bundle["feature_names"]

        features = extract_basic_features(url)

        missing = [
            name for name in feature_names
            if name not in features
        ]

        if missing:
            st.error(
                "Prediction unavailable. The model requires features "
                "that are not currently extracted: "
                + ", ".join(missing)
            )
            st.info(
                "The feature extractor must be aligned with the "
                "training dataset before predictions can be trusted."
            )
        else:
            input_df = pd.DataFrame(
                [[features[name] for name in feature_names]],
                columns=feature_names
            )

            prediction = int(model.predict(input_df)[0])

            if prediction == 1:
                st.error("⚠️ Model result: Potential Phishing")
            else:
                st.success("Model result: Likely Legitimate")

            st.warning(
                "Prototype only. This result does not prove a website "
                "is safe or malicious. Do not enter passwords or "
                "sensitive information based on this result."
            )

    except FileNotFoundError:
        st.error(
            "Model file not found. Check that the model exists "
            "inside the project's model folder."
        )
    except ValueError as error:
        st.error(str(error))
    except Exception as error:
        st.error(f"Something went wrong: {error}")

st.divider()
st.caption("PhishGuard | Educational ML Mini Project")
```
