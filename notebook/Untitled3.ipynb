```python
import pandas as pd
import os
from ucimlrepo import fetch_ucirepo

# Download the official phishing dataset
phishing_data = fetch_ucirepo(id=327)

# Combine input features and target labels
df = pd.concat(
    [
        phishing_data.data.features,
        phishing_data.data.targets
    ],
    axis=1
)

# Find the dataset folder
if os.path.isdir("../dataset"):
    save_path = "../dataset/phishing_dataset.csv"
elif os.path.isdir("dataset"):
    save_path = "dataset/phishing_dataset.csv"
else:
    os.makedirs("dataset", exist_ok=True)
    save_path = "dataset/phishing_dataset.csv"

# Save dataset as a CSV file
df.to_csv(save_path, index=False)

print("Dataset downloaded successfully!")
print("Dataset saved at:", save_path)
print("Dataset shape:", df.shape)
```
